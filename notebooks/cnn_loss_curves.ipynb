# CNN loss curves

Train vs validation loss per epoch for the 1D CNN, read from `results/cnn_loss_history.csv`
(written by `python -m scripts.train_cnn`). The figure is also saved to `results/cnn_loss_curves.png`
for the report.

```python
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

# Jupyter and VS Code run notebooks from the notebook's own folder, so the repo root is one level up.
ROOT = Path("..")
HISTORY_PATH = ROOT / "results/cnn_loss_history.csv"
FIGURE_PATH = ROOT / "results/cnn_loss_curves.png"
```

```python
history = pd.read_csv(HISTORY_PATH)
best = history.loc[history.val_loss.idxmin()]
history
```

```python
plt.plot(history.epoch, history.train_loss, label="Train loss")
plt.plot(history.epoch, history.val_loss, label="Validation loss")
plt.axvline(best.epoch, color="gray", linestyle="--", label=f"Best epoch ({best.epoch:.0f})")
plt.xlabel("Epoch")
plt.ylabel("Binary cross-entropy loss")
plt.title("1D CNN training and validation loss")
plt.legend()
plt.savefig(FIGURE_PATH, dpi=200, bbox_inches="tight")
plt.show()
```

**Reading it:** train loss falls the whole time, while validation loss bottoms out at epoch 12 and then
rises. That gap is overfitting: after epoch 12 the model memorises the training records instead of
learning patterns that carry over to new patients. `train_cnn` keeps only the best-validation
checkpoint (simple early stopping), so the evaluated model is the one from epoch 12.
