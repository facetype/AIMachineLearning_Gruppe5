```python
import pandas as pd
import wfdb
import matplotlib.pyplot as plt
print("ferdig")
```

```python
sti = "../data/raw/ptb-xl-a-large-publicly-available-electrocardiography-dataset-1.0.3/"
df = pd.read_csv(sti + "ptbxl_database.csv", index_col="ecg_id")
print(df.shape)
print(df[["age", "sex", "scp_codes", "strat_fold"]].head())

signal, info = wfdb.rdsamp(sti + df.iloc[0]["filename_lr"])
print(signal.shape)

plt.plot(signal[:, 0])
plt.title(info["sig_name"][0])
plt.show()
```
